```python
import pandas as pd
import numpy as np

df = pd.read_excel("C:/Users/bouto/OneDrive/Documents/gauche_Said_taches.xlsx", sheet_name="Données", header=[0, 1])

for tache in df.columns.get_level_values(0).unique():
    x, y, z = (df[tache][c].dropna() for c in ["X", "Y", "Z"])
    enmo = np.maximum(0, np.sqrt(x**2 + y**2 + z**2) - 1)
    print(tache, enmo.mean())
```
